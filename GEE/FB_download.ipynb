```python
import os
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm import tqdm
import geemap
import ee

def cut_tiles_and_download(roi, output_dir, feature_bands, scale, max_workers):
    coords = roi.bounds().coordinates().getInfo()[0]
    minLon, minLat = coords[0]
    maxLon, maxLat = coords[2]

    # 经纬度换算（粗略）
    tile_size_deg = 2000 / 111320

    tiles = []
    lat = minLat
    while lat < maxLat:
        lon = minLon
        while lon < maxLon:
            lon2 = min(lon + tile_size_deg, maxLon)
            lat2 = min(lat + tile_size_deg, maxLat)
            tiles.append(ee.Geometry.BBox(lon, lat, lon2, lat2))
            lon += tile_size_deg
        lat += tile_size_deg

    print(f"将切成 {len(tiles)} 个切片进行下载...")

    os.makedirs(output_dir, exist_ok=True)

    def export_tile(i_tile):
        i, tile = i_tile
        filename = os.path.join(output_dir, f"tile_{i+1}.tif")
        print(f"开始导出 {filename} ...")
        geemap.ee_export_image(
            feature_bands,
            filename=filename,
            scale=scale,
            region=tile,
            file_per_band=False
        )
        print(f"完成导出 {filename}")
        return filename

    # 使用多线程和tqdm进度条
    results = []
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {executor.submit(export_tile, item): item for item in enumerate(tiles)}
        for f in tqdm(as_completed(futures), total=len(futures), desc="导出切片进度"):
            results.append(f.result())

    return results
```
