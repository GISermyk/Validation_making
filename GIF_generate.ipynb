```python
import os
import numpy as np
import rasterio
import matplotlib.pyplot as plt
from io import BytesIO
from PIL import Image

# 路径
input_folder = r'H:\ICESat-2\GIF_generate\Morecambe'
output_gif = r'H:\ICESat-2\GIF_generate\Morecambe_output_.gif'
basemap_path = r'H:\ICESat-2\Region4_map\Morecambe_map\A2022\Morecambe_2022_FB6_new2.tif'

# 时间
years = ['2019', '2020', '2021', '2022', '2023', '2024']

# 读取底图（只读一次）
with rasterio.open(basemap_path) as base_src:
    base = base_src.read(3)

    # 裁剪（⚠️必须和数据一致）
    base = base[700:2100, 500:1900]

    # 归一化到0-1
    base = base.astype(float)
    base = (base - np.nanmin(base)) / (np.nanmax(base) - np.nanmin(base))

# 读取tif
tif_files = sorted([f for f in os.listdir(input_folder) if f.endswith('.tif')])

frames = []

for i, tif in enumerate(tif_files):
    filepath = os.path.join(input_folder, tif)

    with rasterio.open(filepath) as src:
        data = src.read(1)

        # 裁剪
        data = data[700:2100, 500:1900]

        # 去掉0值（透明）
        data[data == 0] = np.nan

        fig, ax = plt.subplots(figsize=(6, 6))
        ax.axis('off')

        # 底图（灰度）
        ax.imshow(base, cmap='gray')

        # elevation叠加
        cmap = plt.cm.terrain
        cmap.set_bad(color='none')  # NaN透明

        im = ax.imshow(
            data,
            cmap=cmap,
            vmin=-2,
            vmax=5,
            alpha=1
        )

        # 时间
        ax.text(
            0.05, 0.90,
            years[i],
            transform=ax.transAxes,
            fontsize=20,
            color='white',
            weight='bold',
            bbox=dict(facecolor='black', alpha=0.5, pad=3)
        )

        # # colorbar
        cbar = plt.colorbar(im, ax=ax, orientation='horizontal', fraction=0.066, pad=0.02)
        cbar.set_label('Elevation (m)')
        

        # 保存帧
        buf = BytesIO()
        plt.savefig(buf, format='png', bbox_inches='tight', pad_inches=0)
        buf.seek(0)

        img = Image.open(buf)
        frames.append(img.copy())

        buf.close()
        plt.close()

# 输出GIF
frames[0].save(
    output_gif,
    save_all=True,
    append_images=frames[1:],
    duration=1500,
    loop=0
)
```

```python
import scipy.io as sio
import rasterio
from rasterio.transform import from_origin

path = 'G:\JS_DEM_zhang/2016.mat'
out_path = 'G:\JS_DEM_zhang/DEM2016.tif'
# 读取 pathpath.mat 文件
mat_data = sio.loadmat(path)

# 假设数据存在变量名 "data"
data = mat_data["e"]   # 这里 "data" 改成你实际的变量名

# 设置地理信息（需要你知道原始数据的分辨率和左上角坐标）
transform = from_origin(0, 0, 30, 30)  # (左上角X, 左上角Y, 分辨率X, 分辨率Y)

# 保存为 GeoTIFF
with rasterio.open(
    out_path,
    "w",
    driver="GTiff",
    height=data.shape[0],
    width=data.shape[1],
    count=1,
    dtype=data.dtype,
    crs="EPSG:4326",  # 坐标系，根据需要修改
    transform=transform,
) as dst:
    dst.write(data, 1)
```

```python
import scipy.io as sio
import rasterio
from rasterio.transform import from_origin
import numpy as np

# 读取 data.mat
data_file = sio.loadmat('G:\JS_DEM_zhang/2016.mat')
data = data_file["e"]

# 读取 xy.mat
xy_file = sio.loadmat("G:\JS_DEM_zhang/xy.mat")
x = xy_file["x"].squeeze()
y = xy_file["y"].squeeze()

# 假设 x 是东西方向，y 是南北方向，且等间距
res_x = (x[-1] - x[0]) / (len(x) - 1)
res_y = (y[0] - y[-1]) / (len(y) - 1)  # 注意方向

# 仿射变换 (左上角坐标, 分辨率)
transform = from_origin(x[0], y[0], res_x, -res_y)
out_path = 'G:\JS_DEM_zhang/DEM_2016.tif'
# 保存为 GeoTIFF
with rasterio.open(
    out_path,
    "w",
    driver="GTiff",
    height=data.shape[0],
    width=data.shape[1],
    count=1,
    dtype=data.dtype,
    crs="EPSG:4326",  # 如果确定是经纬度，否则可先不写
    transform=transform,
) as dst:
    dst.write(data, 1)
```

```python
import scipy.io as sio
import rasterio
from rasterio.transform import from_origin
import numpy as np

for year in [2015, 2016, 2017, 2018, 2019, 2020]:

    print(year)
# 读取 data.mat
    data_file = sio.loadmat(f'G:\JS_DEM_zhang/{year}.mat')
    data = data_file["e"]
    out_path = f'G:\JS_DEM_zhang/DEM_{year}.tif'
    # nishi针旋转 90 度
    data_rot = np.rot90(data, k=1) 

    # 读取 xy.mat
    xy_file = sio.loadmat("G:\JS_DEM_zhang/xy.mat")
    x = xy_file["x"].squeeze()
    y = xy_file["y"].squeeze()

    # 计算分辨率
    res_x = (x[-1] - x[0]) / (len(x) - 1)
    res_y = (y[0] - y[-1]) / (len(y) - 1)

    # 仿射变换
    transform = from_origin(x[0], y[0], res_x, -res_y)

    # 保存 GeoTIFF
    with rasterio.open(
        out_path,
        "w",
        driver="GTiff",
        height=data_rot.shape[0],
        width=data_rot.shape[1],
        count=1,
        dtype=data_rot.dtype,
        crs="EPSG:4326",  # 如果坐标系未知，可以先不写
        transform=transform,
    ) as dst:
        dst.write(data_rot, 1)
```
